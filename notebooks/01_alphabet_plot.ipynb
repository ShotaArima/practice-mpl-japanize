# Alphabet Plot Baseline

This notebook is the baseline for checking that `uv`, Jupyter, NumPy, and Matplotlib can render ordinary alphabet-only plots before testing Japanese fonts.

```python
import sys

import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print(f"Python: {sys.version}")
print(f"Matplotlib: {matplotlib.__version__}")
print(f"NumPy: {np.__version__}")
print(f"Backend: {matplotlib.get_backend()}")
```

```python
plt.rcParams.update({
    "figure.figsize": (8, 4.5),
    "axes.grid": True,
    "axes.spines.top": False,
    "axes.spines.right": False,
})
```

```python
x = np.linspace(0, 2 * np.pi, 200)
sin_y = np.sin(x)
cos_y = np.cos(x)

fig, ax = plt.subplots()
ax.plot(x, sin_y, label="Sine wave", linewidth=2)
ax.plot(x, cos_y, label="Cosine wave", linewidth=2)
ax.set_title("Alphabet-only Line Chart")
ax.set_xlabel("Angle in radians")
ax.set_ylabel("Amplitude")
ax.legend()
plt.show()
```

```python
labels = ["Alpha", "Beta", "Gamma", "Delta"]
values = [12, 19, 7, 15]

fig, ax = plt.subplots()
ax.bar(labels, values, color=["#4C78A8", "#F58518", "#54A24B", "#B279A2"])
ax.set_title("Alphabet-only Bar Chart")
ax.set_xlabel("Category")
ax.set_ylabel("Value")
plt.show()
```
